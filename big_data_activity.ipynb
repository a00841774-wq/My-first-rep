## c. Missing values

I count missing values per column with `isna().sum()`, then fill the numeric columns with their **median** using `fillna()`.

```python
df.isna().sum()
```

I fill with the **median** rather than the mean because the numeric columns are skewed - a few luxury and sports cars have very high price, horsepower and weight, which drags the mean upward. The median sits in the middle of the data and is not distorted by those outliers, so it is a safer stand-in value.

```python
num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
df.isna().sum()
```

**Reading it:** Eight numeric columns had gaps (Horsepower and Weight 5 each, EngineSize and Cylinders 3 each, MPG_City, MPG_Highway and Wheelbase 2 each, Length 1) - 23 missing cells in total. After the median fill, `isna().sum()` is 0 everywhere, so the dataset is complete.

## d. Duplicate rows

I check for duplicate rows with `duplicated()` and remove any with `drop_duplicates()`.

```python
print('Duplicate rows:', df.duplicated().sum())
before = df.shape[0]
df = df.drop_duplicates()
print('Rows before:', before, '-> Rows after:', df.shape[0])
```

**Reading it:** `duplicated().sum()` is 0, so there are no repeated rows. `drop_duplicates()` removes nothing and the row count stays 428 - I run it anyway to show the check was done.

## f. Selecting columns and filtering rows

I select a few columns, then filter rows two ways: SUVs, and cars with Horsepower above 250.

## g. New calculated variables

I create `Markup` (MSRP minus Invoice) and `Power_to_Weight` (Horsepower divided by Weight).

```python
df['Markup'] = df['MSRP'] - df['Invoice']
df['Power_to_Weight'] = df['Horsepower'] / df['Weight']
df[['Make','Model','MSRP','Invoice','Markup','Horsepower','Weight','Power_to_Weight']].head()
```

**Reading it:** `Markup` is the gap between sticker price and dealer invoice - roughly the dealer margin, and it grows with price. `Power_to_Weight` is horsepower per pound; a higher value means a car is quick for its weight, so light sports cars score highest and heavy SUVs lowest.

## h. Unit conversions

I convert MSRP from USD to Mexican pesos, MPG_City to litres/100km, and Weight from pounds to kilograms.

```python
usd_to_mxn = 18.5  # assumed exchange rate
df['MSRP_MXN'] = df['MSRP'] * usd_to_mxn
df['L_per_100km_City'] = 235.215 / df['MPG_City']
df['Weight_kg'] = df['Weight'] * 0.453592
df[['Model','MSRP','MSRP_MXN','MPG_City','L_per_100km_City','Weight','Weight_kg']].head()
```

**Reading it:** Prices are now in pesos using an assumed rate of 18.5 MXN per USD (easy to update if the rate moves). L/100km is the inverse of MPG, so here a lower number means better fuel economy, and weight is in kilograms for the metric system.

## i. Conditional + for loop: price bands

I write a conditional that labels a car Economy, Mid or Premium by MSRP, then loop over every row to tag the whole dataset.

```python
def price_band(msrp):
    if msrp < 20000:
        return 'Economy'
    elif msrp <= 40000:
        return 'Mid'
    else:
        return 'Premium'

labels = []
for _, row in df.iterrows():
    labels.append(price_band(row['MSRP']))

df['Price_Band'] = labels
df['Price_Band'].value_counts()
```

```python
df[['Make','Model','MSRP','Price_Band']].head(10)
```

**Reading it:** The `for` loop runs the conditional on each of the 428 cars and stores the label. The counts show how the market splits into Economy (under 20,000), Mid (20,000-40,000) and Premium (over 40,000), with most cars in the economy and mid bands and fewer in Premium.
